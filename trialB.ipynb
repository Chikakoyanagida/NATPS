```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
def get_ham(x, e, x0, Vc):
    H = np.zeros((2,2), dtype=complex)
    H[0,0] = e/x0**2 * (x+x0)**2
    H[0,1] = Vc
    H[1,0] = Vc
    H[1,1] = e/x0**2 * (x-x0)**2
    return H

energy_0 = []
energy_1 = []
sample_x = np.linspace(-3,3, 600)

for x in sample_x:
    Ham = get_ham(x, e=0.05, x0=1.0, Vc = 0.05*0.4)
    vals = np.linalg.eigvalsh(Ham)/0.05
    lam_0, lam_1 = vals
    energy_0.append(lam_0)
    energy_1.append(lam_1)

plt.figure()
plt.plot(sample_x, energy_0)
plt.plot(sample_x, energy_1)
plt.show()
```
