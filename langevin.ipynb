```python
from analytical import *
from electronic import *
from engine import *
from integrator import *
from selector import *
from trajectory import *
from path_sampling import *
import numpy as np
from matplotlib import pyplot as plt
```

```python
model = DoubleHarmonic(k=60.8, x0=0.073, vc=0.03)

coords = np.linspace(-0.2,0.2,100)

V0_list = []
V1_list = []
for coord in coords:
    V0 = model.V0(q=coord)
    V0_list.append(V0)
    V1 = model.V1(q=coord)
    V1_list.append(V1)

V0_list = np.array(V0_list)/0.03
V1_list = np.array(V1_list)/0.03

fig, ax = plt.subplots(figsize=(3.4, 2.6), dpi=400)

ax.plot(coords, V0_list)
ax.plot(coords, V1_list)
ax.set_ylim([-1, 20])
ax.grid(True, linestyle=':', alpha=0.5, zorder=0)
plt.show()
```
